```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier(random_state=0)
cross_val_score(clf, data.data, data.target, cv=10)
```

```python
X = data.data
y = data.target
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
from sklearn.metrics import accuracy_score
y_pred = [0, 2, 1, 3]
y_true = [0, 1, 2, 3]
accuracy_score(y_true, y_pred)
accuracy_score(y_true, y_pred, normalize=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt
breast_X_train = breast_cancer.data[:-20]
breast_y_train = breast_cancer.target[:-20]
breast_X_test = breast_cancer.data[-20:]
breast_y_test = breast_cancer.target[-20:]

from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier().fit(breast_X_train, breast_y_train)
```

```python
y_preds = clf.predict(X_test)
y_preds
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_preds, target_names=data.target_names))
```

```python
y_pred = clf.predict(breast_X_test)
accuracy = accuracy_score(breast_y_test, y_pred)
print(accuracy)
```

```python
plt.figure(figsize=(8, 5))
labels = ['Malignant (Ganas)', 'Benign (Jinak)']
counts = [np.sum(breast_cancer.target == 0), np.sum(breast_cancer.target == 1)]
plt.bar(range(len(counts)), counts, tick_label=labels)
plt.show()
```

```python
feature_importance = clf.feature_importances_
indices = np.argsort(feature_importance)[::1]
top_5_features = [breast_cancer.feature_names[i] for i in indices[:5]]
print(top_5_features)
```
